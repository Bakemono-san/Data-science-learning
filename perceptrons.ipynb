Le perceptron simple est un algorithme de classification binaire qui apprend à séparer les données en deux classes à l'aide d'une fonction linéaire. Il est considéré comme l'un des premiers modèles de réseau de neurones artificiels.

Il consiste a prendre des données d'entrée (X) , apprendre une relation et puis faire une prédiction (Y) .

Les etapes de l'algorithme du perceptron simple sont les suivantes :
1. Initialiser les poids (w) et le biais (b) à des valeurs aléatoires.
2. Pour chaque échantillon d'entraînement, calculer le z(la pré-activation) en utilisant la formule : z = w * X + b
3. Appliquer la fonction d'activation (a) : a = f(z) , où f est une fonction d'activation (par exemple, relu ou la fonction sigmoïde). avec sigmoïde : a = 1 / (1 + exp(-z)) , si a est positive alors la classe est 1 sinon la classe est 0.
4. Mettre à jour les poids et le biais en fonction de l'erreur de prédiction : 
   - Si la prédiction est correcte, ne rien faire.
   - Si la prédiction est incorrecte, mettre à jour les poids et le biais en utilisant la règle de mise à jour :
     - calculer l'erreur : error = y_true - a
     - w = w + lr * error * x
     - b = b + lr * error


```python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0

        for _ in range(self.epochs):
            for xi, yi in zip(X, y):
                z = np.dot(self.w, xi) + self.b
                y_pred = 1 if z >= 0 else 0
                
                update = self.lr * (yi - y_pred)
                self.w += update * xi
                self.b += update

    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        return (z >= 0).astype(int)
```

The new perceptron simple algorithm use log loss as the loss function, which is defined as:
L(y_true, y_pred) = - (y_true * log(y_pred) + (1 - y_true) * log(1 - y_pred))
- after that we calculate the derivation in order to know how to update the weights and the bias, the derivation is as follows: calculus = dérivé(L) / dérivé(w)
- we calculate the gradient :
gradient = (y_pred - y_true)x
- the gradient descent:
    w = w - lr * gradient
    b = b - lr * (y_pred - y_true)

then the algorithm will be:

```python
import numpy as np

class LogisticRegression:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0

        for _ in range(self.epochs):
            for xi, yi in zip(X, y):

                # forward
                z = np.dot(self.w, xi) + self.b
                y_hat = self.sigmoid(z)
                
                # calculate loss (optional, for monitoring)
                loss = - (yi * np.log(y_hat) + (1 - yi) * np.log(1 - y_hat))
                print(f"Loss: {loss:.4f}")
                
                if loss < 0.3:
                    print("Loss is low, stopping early.")
                    return

                # gradients
                dw = (y_hat - yi) * xi
                db = (y_hat - yi)

                # update
                self.w -= self.lr * dw
                self.b -= self.lr * db

    def predict_proba(self, X):
        z = np.dot(X, self.w) + self.b
        return self.sigmoid(z)

    def predict(self, X):
        probs = self.predict_proba(X)
        return (probs >= 0.5).astype(int)
```
