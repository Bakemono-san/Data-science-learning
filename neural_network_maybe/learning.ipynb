# in this section we will learn about the perceptron algorithm, which is a simple linear classifier that can be used for binary classification tasks. The perceptron algorithm works by finding a hyperplane that separates the two classes in the feature space.

Here is what i know about the Perceptron mono-layer algorithm:
- The perceptron algorithm is a supervised learning algorithm that can be used for binary classification tasks.
- it works by finding a hyperplane that separates the two classes in the feature space.
- The algorithm iteratively updates the weights of the hyperplane based on the misclassified examples in the training data.
- The algorithm continues to update the weights until it finds a hyperplane that correctly classifies all the training examples or until it reaches a maximum number of iterations.

here is the mathematical representation of the perceptron algorithm:
1. Initialize the weights and bias to zero or small random values.
2. For each training example (x_i, y_i):
    a. Compute the output: y_pred = w * xi + b
    b. check the predicted output against the true label:
        - if >= 0 it is classified as class 1
        - if < 0 it is classified as class 0
    c. When we got a misclassified example, we update the weights and bias:
        - w = w + learning_rate * (y_i - y_pred) * x_i
        - b = b + learning_rate * (y_i - y_pred)
3. Repeat steps 2 until convergence or until a maximum number of iterations is reached.

# Python inplementation of the perceptron algorithm

```python
import numpy as np

def perceptron(X, y, learning_rate=0.01, n_iterations=1000):
    n_samples, n_features = X.shape
    # initialize weights and bias
    weights = np.zeros(n_features)
    bias = 0
    predictions = np.full(n_samples, -1)

    # training loop
    for _ in range(n_iterations):
        print(f"Epoch {_+1}/{n_iterations}")
        for idx, x_i in enumerate(X):
            # here is the preactivation step, where we calculate the linear output
            linear_output = np.dot(x_i, weights) + bias
            
            # here we are applying the step function to get the predicted class label
            y_predicted = np.where(linear_output > 0, 1, 0)
            print(f"Sample: {x_i}, Linear Output: {linear_output}, Predicted: {y_predicted}, Actual: {y[idx]}")

            # update weights and bias if the prediction is wrong
            if y_predicted != y[idx]:
                # the update is proportional to the error (y[idx] - y_predicted) and the learning rate
                update = learning_rate * (y[idx] - y_predicted)
                weights += update * x_i
                bias += update
            else:
                predictions[idx] = y_predicted
                # IF ALL SAMPLES CLASSIFIED CORRECTLY, STOP TRAINING
                if np.array_equal(predictions, y):
                    print("All samples classified correctly. Stopping training.")
                    print(f"Final Weights: {weights}, Final Bias: {bias}")
                    print(f"Predictions: {predictions}")
                    print(f"Actual: {y}")
                    return weights, bias

    return weights, bias

# example usage
if __name__ == "__main__":
    # training data (AND logic gate)
    X = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])
    y = np.array([0, 1, 1, 1])

    weights, bias = perceptron(X, y)
```

# Now we will switch to the MLP(multi layer perceptron) algorithm, which is an extension of the perceptron algorithm that can be used for multi-class classification tasks. The MLP algorithm works by using multiple layers of perceptrons to learn more complex decision boundaries in the feature space.

Here is what i know about the MLP algorithm:
- it use multiple layers of perceptrons to learn more complex decision boundaries in the feature space.
- it can be used for multi-class classification tasks.
- it uses a non-linear activation function to introduce non-linearity into the model, which allows it to learn more complex patterns in the data.
- it will have a main layer called the input layer, one or more hidden layers, and an output layer.
- after activation function is applied to the output of each layer, the output of the final layer is used to make predictions.
- when we got an error in the output layer, we will use backpropagation to update the weights of the network.

Here is the mathematical representation of the MLP algorithm:
1. Initialize the weights and biases of the network to small random values.
2. For each training example (x_i, y_i):
    a. Forward pass:
        - Compute the output of each layer by applying the weights and biases to the input and applying the activation function.
             z = w * x + b 
             a = activation_function(z) based on the activation function we choose, we can have different types of MLPs such as sigmoid MLP, ReLU MLP, etc.
             if sigmoid : a = 1 / (1 + exp(-z))
             if ReLU : a = max(0, z)
        - Calculate the loss between the predicted output and the true label using a loss function such as mean squared error or cross-entropy loss.
             if mean squared error : loss = 1/2 * (y_i - a)^2
             if cross-entropy loss : loss = -y_i * log(a) - (1 - y_i) * log(1 - a)
    b. Backpropagation:
        - We calculate the error : error = a - y_i
        - we calculate the sigmoid derivative : sigmoid_derivative = a * (1 - a)
        - we calculate the delta : delta = error * sigmoid_derivative (how responsible the output layer is for the error)
        - then we calculate the gradient of w5 and w6 : dw5 = sigma_derivative * a1 and dw6 = sigmoid_derivative * a2
        - then we update the output weight : w5 = w5 - learning_rate * dw5 and w6 = w6 - learning_rate * dw6
        - then we calculate the delta of the hidden layer : delta_hidden = sigmoid_derivative * w5 * an(1-an)
        - then we update the hidden weight : wn = delta_hidden * x_i
